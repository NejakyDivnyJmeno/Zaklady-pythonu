# Uvodní opakování
## Proměnné a základní datové typy

```python
vek = 17
cena = 199.90
aktivni = True
jmeno = "Jan"

print(vek, type(vek))
print(cena, type(cena))
print(aktivni, type(aktivni))
print(jmeno, type(jmeno))

cislo = int('42')
text = str(3.14)
print(cislo, type(cislo))
print(text, type(text))
```

## Operace matematické a s řetězci

```python
print(10 + 3)
print(10 - 3)
print(10 * 3)
print(10 / 3)
print(10 // 3) # Dělení se zbytkem
print(10 % 3) # Zbytek po dělení
print(10 ** 3)

pozdrav = "Ahoj" + " " + "Honzo"
cara = "=" * 20
print(pozdrav)
print(cara)

minuty = 2
sekundy = 6
print(f"Čas: {minuty}:{sekundy}")
```

## Vstup a výstup

```python
print("Položka 1", "Položka 2", sep=" | ", end="\r\n")

jmeno = input("Zadej jméno: ")
vek = int(input("Zadej věk: "))
print(f"Ahoj {jmeno}, za rok ti bude {vek + 1}.")
```

## Podmínky a logické výrazy

```python
vek = 16
student = True

if vek < 6:
    cena = 0
elif vek <= 18 and student:
    cena = 120
else:
    cena = 180

print(f"Cena: {cena} Kč")
```

## Cykly (for a while)

```python
for i in range(10, 20, 2):
    print(i, end=" ")
print()

pocet = 3
while pocet > 0:
    print(f"Odpočet: {pocet}")
    pocet -= 1
print("Start!")
```

## Podprogramy (funkce)

```python
def je_sude(cislo):
    return cislo % 2 == 0

print(je_sude(8))
print(je_sude(7))
```

## Náhodná čísla

```python
import random as rnd

kostka = rnd.randint(1, 6)
postava = rnd.choice(["Válečník", "Kouzelník", "Lučištník"])
print(f"Hod: {kostka}, Postava: {postava}")
```

## Kolekce dat (seznam, entice, slovník, množina)

```python
# Seznam
barvy = ["modrá", "červená"]
barvy.append("zelená")
print(barvy[0])
print("modrá" in barvy)

# Entice
bod = (10, 20)
x, y = bod
print(f"X: {x}, Y: {y}")

# Slovník
ceny = {"jablko":25, "hruška":30}
ceny["banán"] = 40

for plod, cena in ceny.items():
    print(f"{plod}: {cena} Kč")

# Množina
cisla = [1, 2, 2, 3, 1, 4]
unikatni = set(cisla)
print(f"Unikátní: {unikatni}, počet: {len(unikatni)}")
```
